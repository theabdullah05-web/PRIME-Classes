```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("./thyroid_dataset.csv")
```

```python
df
```

```python
X = df.drop(columns="Outlier_label")
y = df["Outlier_label"]
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest
clf = IsolationForest(n_estimators=200, contamination=0.036,random_state=42)
labels=clf.fit_predict(X_scaled)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)
```

```python
import seaborn as sns
plt.figure(figsize=(8,6))
sns.scatterplot(x=X_pca[:,0],y=X_pca[:,1],c=labels)
```

```python
import numpy as np
normal=np.sum(labels==1)
outliers=np.sum(labels==-1)
print(f"normal: {normal}")
print(f"outliers: {outliers}")
```

# LOF

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
neighbs = LocalOutlierFactor(contamination=0.036)
labels=neighbs.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], c=labels)
```

```python
import numpy as np

normal = np.sum(labels == 1)
outliers = np.sum(labels == -1)
print(f"normal: {normal}")
print(f"outliers: {outliers}")
```
