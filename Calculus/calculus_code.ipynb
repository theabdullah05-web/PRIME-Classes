```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,100,100_000)
y=np.sqrt(x)

plt.plot(x,y, label="y = √x")
plt.grid(True)
plt .title("Calculus Functions Relationship")
plt.xlabel("Values")
plt.ylabel("Square Roots")
plt.legend()
```

```python
x=np.linspace(-2,2,100)
y=x**2
plt.plot(x,y, label='y = sin(x)', color='blue')
plt.grid()
plt.axvline(0, color='black', linewidth=1.5, linestyle='--', label='x = 0')
plt.axhline(0, color='black', linewidth=1.5, linestyle='--', label='y = 0')
plt.title('Graph of y = sin(x)')
plt.legend()
```
