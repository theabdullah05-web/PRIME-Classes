```python
#Real Data
#Oscar Winner Movies
oscar_movies=["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232] #in $M

#Plot
plt.plot(oscar_years,oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
non_oscar_movies=["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years=[2008,2009,2010,2011,2012]
non_oscar_revenue=[378,2788,829,185,275]
plt.plot(oscar_years,oscar_revenue,label="oscar movies")
plt.plot(non_oscar_years,non_oscar_revenue,label="non-oscar movies")
plt.title("Oscar movies vs Non-Oscar movies in terms of Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.savefig("abc1.png")
```

```python
#Bars
#Oscar Winner Movies
oscar_movies=["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232] #in $M

plt.bar(oscar_years,oscar_revenue, color="green")
plt.title("Revenue in each Year for Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
for i in range(len(oscar_years)):
    plt.text(oscar_years[i],oscar_revenue[i]+20,oscar_revenue[i],ha="center")
plt.ylim(0,max(oscar_revenue)+100)
```

```python
#Compare
#Oscar Winner Movies
oscar_movies=["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232] #in $M

#Non-Oscar Winner Movies
non_oscar_movies=["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years=[2008,2009,2010,2011,2012]
non_oscar_revenue=[378,2788,829,185,275]
width=0.4
x=np.arange(len(oscar_years))
plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in M$)")
plt.bar(x-width/2,oscar_revenue,width,label="Oscar Movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="Non-Oscar Movies")
plt.legend()
plt.xticks(x,oscar_years)
```

```python
#Scatter
people=["PersonA","PersonB","PersonC","PersonD","PersonE","PersonF","PersonG","PersonH","PersonI","PersonJ"]
age=[22,25,30,35,40,45,50,55,60,65]
blood_pressure=[110,115,120,122,125,130,135,123,145,150]

colors=["green" if x< 135 else "red" for x in blood_pressure]
plt.scatter(age,blood_pressure,s=blood_pressure,cmap='YlOrRd',c=blood_pressure)
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.title("Blood Pressure with Respect to Age")
plt.grid()           
plt.colorbar(label="BP")
for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],blood_pressure[i]),xytext=(age[i]-3,blood_pressure[i]+2))
plt.ylim(106,max(blood_pressure)+5)
plt.xlim(min(age)-4,max(age)+5)
```

```python
# Multiple Plots

cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp,winter_humidity,label="Winter")
plt.scatter(summer_temp,summer_humidity,label="Summer")
plt.xlabel("Temprature °C")
plt.ylabel("Humidity %")
plt.title("Relation between Humidity and Temprature")
plt.legend()
```

```python
#Pie
expenses = ["Rent", "Groceries", "Utilities", "Transport", "Entertainment"]
amounts = [15000, 8000, 3500, 2000, 1500]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",wedgeprops={
    "edgecolor":"black",
    "linestyle":"--"
},explode=[0,0,0,0,.5],shadow=3)
```

```python
# multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]
plt.hist(fraud_transactions,label="Fraud",color="red",edgecolor="black",bins=20,alpha=0.5)
plt.hist(legit_transactions,label="Legit",color="green",edgecolor="black",bins=20,alpha=0.5)
plt.title("Transaction Amount Distribution - Legit vs Fraud")
plt.xlabel("Amount (in $)")
plt.ylabel("Frequency")
plt.axvline(
    x=300
)
```
