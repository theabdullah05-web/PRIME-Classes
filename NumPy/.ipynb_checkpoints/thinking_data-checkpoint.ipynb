```python
import json;
```

```python
def load_data(file_name):
    with open(file_name,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_users=set()
    for i in data:
        #Clean Ratings
        raw_rating=str(i["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        i["rating"]=raw_rating
        
        #Handle Missing Values
        raw_age=i.get("age")
        if(raw_age==None):
            i["age"]=None

        #Deduplicate data
        if(i["name"].strip() in unique_users):
            continue
        unique_users.add(i["name"])
        cleaned_data.append(i)
    return cleaned_data
```

```python
clean_data(data)
```

```python
#Get Meaningful Insights
def get_insights(data):
    #avg rating
    rating_sum=0
    count=0
    for i in data:
        rating_sum+=float(i["rating"])
        count+=1
    avg=rating_sum/count
    print(avg)

    #percentage of users with poor rating
    count_low=0
    for i in data:
        rating_val=float(i["rating"])
        if(rating_val<3):
            count_low+=1
    print(f"percentage of low rating={count_low}")
```

```python
print(get_insights(data))
```
